# AUTOVAULT AI — Data Exploration

Exploratory analysis of vehicle master dataset, specification attributes, fuel types, and purchase price distributions.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

print("All exploration libraries imported successfully.")
```

```python
from pathlib import Path

# Locate vehicle_master.csv whether running from notebooks/ or repo root
data_path = Path("../data/vehicle_master.csv")
if not data_path.exists():
    data_path = Path("data/vehicle_master.csv")

df = pd.read_csv(data_path)
print(f"Loaded vehicle_master.csv: {df.shape[0]} records, {df.shape[1]} features.")
df.head()
```

```python
# Statistical summary of numeric and categorical attributes
df.describe(include="all")
```

```python
# Price distribution visualization
fig = px.histogram(
    df,
    x="purchase_price_lakh",
    color="fuel_type",
    nbins=20,
    marginal="box",
    title="Vehicle Purchase Price Distribution by Fuel Type (₹ Lakh)",
    template="plotly_white"
)
fig.update_layout(
    xaxis_title="Purchase Price (₹ Lakh)",
    yaxis_title="Count",
    bargap=0.08
)
fig.show()
```
